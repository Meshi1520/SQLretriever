```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
```

```python
sentences = [
    "I love my dog",
    "I love my cat",
    "You love my dog!",
    "Do you think my dog is amazing?"

]
# The labels should match the number of sentences. Assuming binary classification for these 4 sentences.
labels = np.array([1, 1, 0, 1]) # Example labels: 1 for positive, 0 for negative
```

```python
vocab_size = 200
tok = Tokenizer(num_words = vocab_size,oov_token = "<OOV") #oov = out of vocabulary if the word is out of vocab
tok.fit_on_texts(sentences)
seqs = tok.texts_to_sequences(sentences)
maxlen = max(len(s) for s in seqs)
X = pad_sequences(seqs,maxlen = maxlen, padding = 'post') # post = pad after all thw word have been written
y = labels
```

```python
X[0]
```

```python
X[3]
```

```python
from tensorflow.keras.layers import Input
embed_dim = 16
rnn_units = 8
inp = Input(shape =(maxlen,),dtype = "int32",name = 'model_input')
```

```python
x = Embedding(input_dim = vocab_size,output_dim=embed_dim , mask_zero = True,name = 'embed')(inp) #mask zero helps us to padding
```

```python
# The SimpleRNN layer definition has been moved to cell IJzupeocB9bd for better graph connectivity.
```

```python
rnn = SimpleRNN(units = rnn_units, return_sequences= True,return_state = True, name = 'simple_rnn')
x_sequences, x_last = rnn(x)
out = Dense(1,activation = 'sigmoid', name = 'out')(x_last)
```

```python
model = Model(inputs = inp, outputs = out)
```

```python
model.compile(optimizer  = 'adam',loss = 'binary_crossentropy',metrics = ['accuracy'])
```

```python
model.summary()
```

```python
model.fit(X,y ,epochs = 15, batch_size = 8,verbose = 1)
```

```python
intermediate_model = Model(inputs=inp, outputs=x_last)
print(intermediate_model.summary())
```

Now, let's see the output of this intermediate model when given your input data `X`:

```python
intermediate_output = intermediate_model.predict(X)
display(intermediate_output)
print(f"Shape of intermediate output: {intermediate_output.shape}")
```
